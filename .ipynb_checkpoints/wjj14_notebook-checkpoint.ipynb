Introduction: Our project is to choose the best Pittsburgh neighborhood based on traffic. The dataset I chose contains a list of traffic permits that were received for certain areas undergoing construction. We came up with our traffic metric after discussing different ideas, such as crime and walkability. We ultimately settled on traffic because we felt it was a lighthearted topic that is vitally important to a good neighborhood, especially one in Pittsburgh. I am using the Right of Way Permits dataset to determine which neighborhoods have the most traffic construction permits. This will allow me to determine the best neighborhood to live in based on construction around it. 

```python
import pandas as pd
import matplotlib.pyplot as plt

FILENAME = "Permits.csv"
NEIGHBORHOOD_COLUMN = "neighborhood"

data = pd.read_csv("Permits.csv")

counts = data[NEIGHBORHOOD_COLUMN].value_counts()

largest3 = counts.head(3)
smallest3 = counts.tail(3)

plt.figure(figsize=(8, 5))
plt.bar(largest3.index, largest3.values, color="steelblue")
plt.title("3 Neighborhoods with the Most Permits")
plt.xlabel("Neighborhood")
plt.ylabel("Number of Permits")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("largest_permits.png")

plt.figure(figsize=(8, 5))
plt.bar(smallest3.index, smallest3.values, color="indianred")
plt.title("3 Neighborhoods with the Fewest Permits")
plt.xlabel("Neighborhood")
plt.ylabel("Number of Permits")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("smallest_permits.png")

plt.show()
```

Conclusion: I believe, based on the data set I had and the graphs, that the best neighborhood to live in based on traffic is Mount Oliver Borough.
